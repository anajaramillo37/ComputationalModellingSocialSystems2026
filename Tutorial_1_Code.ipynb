```python
# A class is like a blueprint for creating objects (instances). It groups together data (attributes) and functions (methods)
# that operate on that data. In this case, we are creating a class called 'OurAgent' to represent an agent in our model.
class OurAgent:
    # __init__ is the method, called when we create a new object.
    # Each method in a class takes 'self' as the first argument, which refers to the instance of the class that is being created or used.
    def __init__(self, wealth, name):
        print("Creating a new agent")
        # We store the wealth and name as attributes of the agent.
        self.wealth = wealth  # Amount of money the agent has
        self.name = name      # The agent's name
    
    def tell_wealth(self):
        print(f"I am {self.name} and I have {self.wealth} coins!")
    
    # A Method to transfer 1 coin to Agent "to_whom".
    def distribute_wealth(self, to_whom: OurAgent):
        if self.wealth == 0: # If the agent has no wealth, they cannot give any away.
            print("Sorry, I am broke!")
        else:
            # Increase the recipient's wealth by 1
            to_whom.wealth = to_whom.wealth + 1
            # Decrease this agent's wealth by 1
            self.wealth = self.wealth - 1
            print(f"I am {self.name} and I will transfer 1 coin to agent {to_whom.name}")

# create some agents to play around
# Create an agent named Alex with 12 coins
alex = OurAgent(wealth=12, name="Alex")
alex.tell_wealth()

# Create an agent named Frederick with 33 coins, this time not explicitly naming the parameters
# (the order of parameters in the __init__ method is used).
frederick = OurAgent(33, "Frederick")
# Would be the same as frederick = OurAgent(wealth=33, name="Frederick")
frederick.distribute_wealth(alex) # give a coin to Alex
frederick.tell_wealth()
alex.tell_wealth()

# Create Jennifer with 0 coins (broke)
jennifer = OurAgent(0, "Jennifer")
jennifer.tell_wealth()
alex.tell_wealth()
jennifer.distribute_wealth(alex)  # Jennifer tries to give to Alex, but she's broke
jennifer.tell_wealth()
alex.tell_wealth()
```

```python
import mesa
```

```python
# Here we are using a already programmed class "Agent" from the Mesa library. We will extend their basic Agenutwith additional functionality.
# So let's inherit from mesa.Agent.
class OurAgent(mesa.Agent):
    # Mesa also wants us to pass the model as an argument to the __init__ method, so we can access the model from which the agent is created.
    # Also a fix compared to the code in the lecture: In previous Mesa-Versions, you had to specify the unique_id. Mesa now
    # automatically assigns every agent a unique_id, so we don't need to pass it as an argument. Sorry for that
    def __init__(self, model, wealth):
        # We also need to initialize the parent class (mesa.Agent). We do that by calling super().__init__(...).
        super().__init__(model)
        # Inside the __init__ method of mesa, every agent gets a unique_id from mesa which we can access via self.unique_id

        print(f"Creating a new agent with the unique id {self.unique_id}")
        self.wealth = wealth
    
    def tell_wealth(self):
        print(f"I am agent number {self.unique_id} and I have {self.wealth} coins!")
    
#    def distribute_wealth(self, to_whom: OurAgent): # More explicit but depending on the Python version requires "from __future__ import annotations" 
    def distribute_wealth(self, to_whom):
        if self.wealth == 0:
            print("Sorry, I am broke!")
        else:
            to_whom.wealth = to_whom.wealth + 1
            self.wealth = self.wealth - 1
            print(f"I am agent number {self.unique_id} and I will transfer 1 coin to agent number {to_whom.unique_id}")
```

```python
class MoneyModel(mesa.Model):
    def __init__(self):
        super().__init__()

        self.agent_a = OurAgent(model=self, wealth=2)
        self.agent_b = OurAgent(model=self, wealth=0)
        self.agent_c = OurAgent(model=self, wealth=100)

    def step(self):
        self.agent_a.distribute_wealth(self.agent_b)
        self.agent_b.distribute_wealth(self.agent_c)
        self.agent_c.distribute_wealth(self.agent_a)

money_model = MoneyModel()
money_model.step()
money_model.step()
money_model.step()
money_model.step()
```
